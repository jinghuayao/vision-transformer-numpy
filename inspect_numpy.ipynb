# Inspect NumPy ViT experiments 🔍
NumPy port of `inspect.ipynb`: same 5 cells (imports → samples → load → curves → attention). Run `vision_transformers_numpy.ipynb` first so `experiments/vit-numpy-demo/` exists.

```python
from utils_numpy import (
    visualize_images, visualize_attention, load_experiment, plot_metrics,
)
from IPython.display import Image, display

import os
_CANDIDATES = [
    os.environ.get("VIT_DATA_ROOT", ""),
    "/Users/jinghuayao/Downloads/vision-transformer-from-scratch/data",
    "./data",
    "../vision-transformer-from-scratch/data",
]
DATA_ROOT = next(
    (c for c in _CANDIDATES
     if c and os.path.isdir(os.path.join(c, "cifar-10-batches-py"))),
    None,
)
assert DATA_ROOT is not None, "CIFAR-10 pickles not found; set VIT_DATA_ROOT"
print("DATA_ROOT =", DATA_ROOT)
exp_name = "vit-numpy-demo"  # trained by vision_transformers_numpy.ipynb
print("inspecting experiment:", exp_name)
```

```python
# Show some training images (cf. inspect.ipynb cell 2)
visualize_images(data_root=DATA_ROOT, output="samples.png")
display(Image("samples.png"))
```

```python
# Load Experiment (cf. inspect.ipynb cell 3; checkpoint is model_final.npz)
config, model, train_losses, test_losses, accuracies = load_experiment(exp_name)
print("config:", {k: config[k] for k in ("patch_size", "hidden_size", "num_hidden_layers", "num_attention_heads")})
print(f"epochs logged: {len(train_losses)}")
```

```python
# Loss/accuracy curves (cf. inspect.ipynb cell 4)
plot_metrics(train_losses, test_losses, accuracies, output="metrics.png")
display(Image("metrics.png"))
```

```python
# CLS-token attention overlays (cf. inspect.ipynb cell 5)
visualize_attention(model, data_root=DATA_ROOT, output="attention.png")
display(Image("attention.png"))
```
